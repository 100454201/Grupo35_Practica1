## 2_ModeloDummy

Ahora tratamos los valores sabiendo que los valores de Y (Atrittion) estan desbalanceados

```python
#Separamos la variable objetivo con las variables predictorias
X = data.drop(columns=["Attrition"])  
y = data["Attrition"]

#Y ahora dividimos el dataset asegurandonos que la proporción de clases se mantiene igual en los dos conjuntos
# División estratificada en 80% train y 20% test
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)

#Vamos a comprobar los valores
print("Distribución en Train:")
print(y_train.value_counts(normalize=True))

print("\nDistribución en Test:")
print(y_test.value_counts(normalize=True))
```

La proporción de clases en ambos conjuntos es casi idéntica, lo que confirma que stratify=y ha mantenido el balance de clases en la división.

Codificamos la Salida

```python
#Vamos a ver si la salida de Atrittion es binaria
print("Valores únicos del mapeo:", data["Attrition"].unique())
```

Como los valores de Atrittion son 0 y 1 ya, no hace falta codificar la salida

```python
#Analizamos las variables categóricas y numéricas que tenemos
cat_cols = X_train.select_dtypes(include="object").columns
num_cols = X_train.select_dtypes(exclude="object").columns
cat_cols
```

Pipeline de las variables categóricas

```python
#Valor más frecuente para calores nulos
#Codificamos en OneHotEncoder
#PCA
cat_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
    ('pca', PCA(n_components=5))
])
```

Pipeline de las variables numéricas

```python
#Imputa valores nulos usando la media de los 5 vecinos cercanos
#Escala datos de manera robusta frente a valores atipicos
num_transformer = Pipeline(steps=[
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler', RobustScaler())
])
```

Preprocesador

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_transformer, num_cols),
        ('cat', cat_transformer, cat_cols)
    ])
preprocessor
```

Empezaremos con un Modelo Trivial (dummy) como referencia base

```python

from sklearn.dummy import DummyClassifier
from sklearn.metrics import classification_report, balanced_accuracy_score

# Creamos un pipeline igual que los modelos reales
dummy_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', DummyClassifier(strategy='most_frequent'))  # Siempre predice la clase mayoritaria
])

# Entrenamiento
dummy_pipeline.fit(X_train, y_train)

# Predicción
y_pred_dummy = dummy_pipeline.predict(X_test)

# Evaluación
print("Balanced Accuracy Dummy:", balanced_accuracy_score(y_test, y_pred_dummy))
print("\Reporte Dummy:")
print(classification_report(y_test, y_pred_dummy))
```

Con este modelo dummy, lo único que queremos demostrar es que los modelos reales mejoran la línea base. El modelo dummy solo acierta la clase mayoritaria

## 3_Métodos optimizados: KNN y Trees

Clasificador

```python
from sklearn.model_selection import cross_val_score
clf = Pipeline(steps=[('preprocessor', preprocessor),
                      ('classifier', tree.DecisionTreeClassifier(class_weight='balanced'))])
clf
```

Como el dataset está desbalanceado usamos como mejor resultado "balanced_accuracy"

```python
cross_val_score(clf, X_train, y_train, cv=5, scoring="balanced_accuracy").mean()
```

Esto quiere decir que  la métrica que esstamos evaluando (balanced_accuracy) tiene un valor de 0.8326 (83.26%) en promedio en la validación cruzada. 83.67% de precisión balanceada en validación cruzada, lo cual es bueno, considerando que los datos están desbalanceados.

Vamos a buscar los mejores hiperparámetros, probando 150 combinaciones aleatorias para encontrar la mejor (n_iter = 150)

```python
from sklearn.model_selection import RandomizedSearchCV

# Definir distribución de hiperparámetros
num_transformer_dist = {
    'preprocessor__num__imputer__n_neighbors': list(range(2, 15)),
    'preprocessor__num__imputer__add_indicator': [True, False]
}

cat_transformer_dist = {
    'preprocessor__cat__imputer__strategy': ['most_frequent', 'constant'],
    'preprocessor__cat__imputer__add_indicator': [True, False],
    'preprocessor__cat__pca__n_components': list(range(2, 5))
}

random_forest_dist = {
    'classifier__max_depth': list(range(2, 20)),  
    'classifier__min_samples_split': list(range(20, 200)),  
    'classifier__max_features': ['sqrt', 'log2', None]  # Tamaño de submuestra de features
}

# Unir todos los hiperparámetros en un solo diccionario
param_dist = {**num_transformer_dist, **cat_transformer_dist, **random_forest_dist}

# Definir búsqueda de hiperparámetros
random_search = RandomizedSearchCV(
    clf, 
    param_distributions=param_dist, 
    n_iter=150, 
    cv=5,  # Validación cruzada con 5 folds
    scoring='balanced_accuracy',  # Optimizamos precisión balanceada
    random_state=42,  
    n_jobs=-1  # Paralelización para mayor velocidad
)

# Ejecutar la búsqueda de hiperparámetros
random_search.fit(X_train, y_train)

# Mostrar los mejores hiperparámetros encontrados
print("Mejores hiperparámetros:", random_search.best_params_)
print("Mejor balanced accuracy:", random_search.best_score_)

```

Antes (cross_val_score) → ~83.67%
Evaluamos el modelo por defecto de DecisionTreeClassifier con validación cruzada (cv=5).
    No habiamos optimizado hiperparámetros, solo usamos los valores predeterminados.

Ahora (RandomizedSearchCV) → ~77.83%
Se optimizaron los hiperparámetros, lo que puede haber cambiado la generalización del modelo.
La validación cruzada dentro de RandomizedSearchCV usa diferentes particiones de los datos.
Algunos hiperparámetros pueden haber hecho que el modelo sea menos flexible (menor overfitting), reduciendo la accuracy, pero no por eso es peor modelo. De hecho es probable que sea más robusto y generalizable.
Nota: El segundo código es una versión ""mejorada"" por ChatGpt al código dado en el Tutorial2

## SVM Lineal (kernel = 'linear')

```python
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
# Pipeline con SVM lineal
svm_linear_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', SVC(kernel='linear', class_weight='balanced'))
])
```

Búsqueda de hiperparámetros con C
Cuanto más grande es C, menos tolerancia a errores tiene el modelo (puede sobreajustar).
Cuanto más pequeño es C, más ancho es el margen, y se puede generalizar mejor.

```python
from sklearn.model_selection import RandomizedSearchCV

svm_linear_param_dist = {
    'classifier__C': [0.01, 0.1, 1, 10, 100]
}
svm_linear_search = RandomizedSearchCV(
    svm_linear_pipeline,
    param_distributions=svm_linear_param_dist,
    n_iter=5,  # Solo hay 5 opciones para C, así que probamos todos
    scoring='balanced_accuracy',
    cv=5,
    n_jobs=-1,
    random_state=42
)

svm_linear_search.fit(X_train, y_train)

# Resultados
print("Mejor C:", svm_linear_search.best_params_)
print("Mejor balanced accuracy (CV):", svm_linear_search.best_score_)

```

Ahora evaluaremos el modelo SVM lineal en el test set (X_test) y visualizaremos el espacio de decisión

```python
from sklearn.metrics import classification_report, balanced_accuracy_score

# Predecir con el mejor modelo encontrado
y_pred_svm = svm_linear_search.predict(X_test)

# Mostrar resultados
print("Balanced Accuracy en test:", balanced_accuracy_score(y_test, y_pred_svm))
print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred_svm))
```

El modelo identifica bien los abandonos (recall = 0.72), aunque a costa de una precisión más baja (0.35). Tiene menos precisión que el anterior modelo. Esto se puede deber a que los datos estań solapados, por lo que un modelo lineal no es lo mejor.
Para el negocio, detectar abandonos (aunque con algún falso positivo) puede ser más importante que evitarlos todos.
Muy buen equilibrio para tratar un problema desbalanceado.

Vamos a visualizar los datos ahora

```python

from sklearn.decomposition import PCA
from sklearn.svm import SVC
import matplotlib.pyplot as plt
import numpy as np

#Aplicar preprocesamiento a los datos completos
X_train_proc = preprocessor.fit_transform(X_train)
X_test_proc = preprocessor.transform(X_test)

# Reducir dimensionalidad a 2D con PCA
pca = PCA(n_components=2)
X_train_2D = pca.fit_transform(X_train_proc)
X_test_2D = pca.transform(X_test_proc)

#Entrenar un SVM lineal solo para visualización (C=10 porque fue el óptimo)
svm_vis = SVC(kernel='linear', C=10, class_weight='balanced')
svm_vis.fit(X_train_2D, y_train)

#Gráfico 1: Puntos del dataset en 2D
plt.figure(figsize=(6, 4))
plt.scatter(X_train_2D[y_train == 0, 0], X_train_2D[y_train == 0, 1], c='blue', label='No Abandona')
plt.scatter(X_train_2D[y_train == 1, 0], X_train_2D[y_train == 1, 1], c='red', label='Abandona')
plt.title("Datos SVM lineal en 2D (PCA)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

#Gráfico 2: Frontera de decisión
x_min, x_max = X_train_2D[:, 0].min() - 1, X_train_2D[:, 0].max() + 1
y_min, y_max = X_train_2D[:, 1].min() - 1, X_train_2D[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 500),
                     np.linspace(y_min, y_max, 500))

Z = svm_vis.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.figure(figsize=(6, 4))
plt.contourf(xx, yy, Z, cmap=plt.cm.bwr, alpha=0.3)
plt.scatter(X_train_2D[y_train == 0, 0], X_train_2D[y_train == 0, 1], c='blue', s=20, label='No Abandona')
plt.scatter(X_train_2D[y_train == 1, 0], X_train_2D[y_train == 1, 1], c='red', s=20, label='Abandona')
plt.title("Frontera de decisión SVM lineal (PCA 2D)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Cada punto representa un empleado proyectado en los dos primeros componentes principales.

Se observa una cierta separación entre clases, pero también solapamiento, lo que explica la dificultad del modelo.

    El SVM lineal ha trazado un hiperplano (línea recta) que separa las clases en el espacio 2D.

    Puedes ver claramente la zona de predicción roja (abandona) y la zona azul (no abandona).

Esto confirma gráficamente que el modelo se esfuerza por separar ambas clases con una frontera lineal, aunque hay solapamiento debido a la complejidad del problema.

## SVM Radial (kernel = 'rbf')

```python
svm_rbf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', SVC(kernel='rbf', class_weight='balanced'))
])
svm_rbf_param_dist = {
    'classifier__C': [0.01, 0.1, 1, 10, 100],
    'classifier__gamma': [0.001, 0.01, 0.1, 1]
}

```

```python
from sklearn.model_selection import RandomizedSearchCV

svm_rbf_search = RandomizedSearchCV(
    svm_rbf_pipeline,
    param_distributions=svm_rbf_param_dist,
    n_iter=15,  
    scoring='balanced_accuracy',
    cv=5,
    n_jobs=-1,
    random_state=42
)

# Entrenamiento con búsqueda de hiperparámetros
svm_rbf_search.fit(X_train, y_train)

# Mostrar mejores resultados
print("Mejores hiperparámetros SVM RBF:", svm_rbf_search.best_params_)
print("Mejor balanced accuracy (CV):", svm_rbf_search.best_score_)
```

Muy superior al SVM lineal (≈73%)

```python
from sklearn.metrics import classification_report, balanced_accuracy_score

# Predecir con el mejor modelo RBF encontrado
y_pred_rbf = svm_rbf_search.predict(X_test)

# Mostrar resultados
print("Balanced Accuracy en test:", balanced_accuracy_score(y_test, y_pred_rbf))
print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred_rbf))
```
