# Predicciones Modelo Final

```python
import joblib 
# Cargar el modelo entrenado desde el archivo .pkl
modelo_final = joblib.load("../modelo_final.pkl")
```

```python
import pandas as pd

# Cargar datos de competición
competition = pd.read_csv("../data/attrition_competition_35.csv.gz", compression="gzip")
print("Dimensiones del conjunto de competición:", competition.shape)
```

```python
# Realizar predicciones sobre el conjunto de competición
y_pred_competition = modelo_final.predict(competition)
```

```python
# Guardar predicciones en un archivo llamado predicciones.csv
pd.DataFrame({
    "Prediction": y_pred_competition
}).to_csv("../predicciones.csv", index=False)
```
